```python
import os
import pandas as pd

# Tabel Data
file_name = 'andrew.xlsx'

# Deteksi lokasi file andrew.xlsx secara fleksibel (direktori aktif, folder notebook, Colab, dll)
possible_paths = [
    file_name,
    os.path.join(os.getcwd(), file_name),
    os.path.join(r'd:\Tugas_MTK', file_name),
    f'/content/{file_name}',
    os.path.join(os.path.dirname(os.path.abspath('__file__')) if '__file__' in globals() else '', file_name)
]

path = next((p for p in possible_paths if p and os.path.exists(p)), file_name)

print(f'Memuat data dari: {path}')
dataraw = pd.read_excel(path)
display(dataraw)
```

```python
# Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
display(datafrq)
```

```python
import matplotlib.pyplot as plt

# Grafik Garis
datafrq.plot(kind='line', figsize=(10, 6), marker='o', color='skyblue')
plt.title('Grafik Garis Frekuensi Grade', fontsize=16)
plt.xlabel('Grade', fontsize=12)
plt.ylabel('Frekuensi', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Grafik Batang
datafrq.plot(kind='bar', figsize=(10, 6), color='lightcoral')
plt.title('Grafik Batang Frekuensi Grade', fontsize=16)
plt.xlabel('Grade', fontsize=12)
plt.ylabel('Frekuensi', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pie Chart
fig, ax = plt.subplots(figsize=(8, 8))
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.1f%%', legend=False, ax=ax, startangle=90, colors=plt.cm.Paired.colors)
ax.set_title('Distribusi Frekuensi Grade', fontsize=16)
ax.set_ylabel('') # Hide y-label for pie chart
plt.tight_layout()
plt.show()
```

```python
# Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
display(stats)
```
